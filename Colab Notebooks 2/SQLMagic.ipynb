3.Connecting to DuckDB 
如果你有一个数据库文件，比如 nyc_data.db，
你就可以把这里的连接字符串从 :memory:
改成指向那个 .db 文件的路径。不需要再自己创建表

给你一个 zip 文件，里面装着一堆CSV / GeoJSON / shapefile
我们需要先在内存里创建一个空数据库
然后对 zip 里的每一个文件都创建一个表

```python
%sql duckdb:///:memory:
# %sql duckdb:///path/to/file.db
```
